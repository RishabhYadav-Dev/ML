**# Implementing Bagging (BaggingClassifier)**

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# Feature selection and train-test split
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
# Note: If your CSV includes an 'Id' column as the first column, use: X = df.iloc[:, 1:-1]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Implementing Bagging (BaggingClassifier)
# For simplicity, we are going to use as base estimator a Decision Tree with fixed parameters.
# Using max_depth=2 to prevent overfitting on this relatively simple dataset.
tree_bagging = DecisionTreeClassifier(max_depth=2, random_state=42)

# The bagging ensemble classifier is initialized with:
# estimator = DecisionTree (updated from base_estimator)
# n_estimators = 10 : number of base estimators in the ensemble
# max_samples = 0.8 : it's gonna be taken randomly 80% of items with replacement
# bootstrap = True : means that the sampling is gonna be with replacement
bagging_iris = BaggingClassifier(estimator=tree_bagging, n_estimators=10, max_samples=0.8, bootstrap=True, random_state=42)

# Training
bagging_iris.fit(X_train, y_train)

# Evaluating
print(f"Bagging Train score: {bagging_iris.score(X_train, y_train):.4f}")
print(f"Bagging Test score: {bagging_iris.score(X_test, y_test):.4f}")
```

**# Implementing Boosting (AdaBoostClassifier)**

```python
from sklearn.ensemble import BaggingClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier

# Implementing Boosting (AdaBoostClassifier)
# The base learner will be a decision tree with depth = 2
tree_adaboost = DecisionTreeClassifier(max_depth=2, random_state=42)

# AdaBoost initialization
# It's defined the decision tree as the base learner (updated from base_estimator)
# The number of estimators will be 10
# The learning rate controls the contribution of each weak learner
adaboost_iris = AdaBoostClassifier(estimator=tree_adaboost, n_estimators=10, learning_rate=0.1, random_state=42)

# Train!
adaboost_iris.fit(X_train, y_train)

# Evaluation
print(f"AdaBoost Train score: {adaboost_iris.score(X_train, y_train):.4f}")
print(f"AdaBoost Test score: {adaboost_iris.score(X_test, y_test):.4f}")
```

Aim:
To implement Bagging (BaggingClassifier) and Boosting (AdaBoostClassifier) ensemble techniques in Python using the Iris dataset.

Objective:
To build ensemble classification models using Decision Trees as base learners, evaluate their training and testing accuracies, and understand how parallel (Bagging) and sequential (Boosting) learning reduce variance and bias.

Theory:
Ensemble Learning is a machine learning paradigm where multiple weak learners (base models, such as shallow Decision Trees) are combined to solve the same problem and achieve higher predictive accuracy and stability than any single model alone.

1. Bagging (Bootstrap Aggregating):
- Concept: Bagging trains multiple base models independently and in parallel on different random subsets of the training data.
- Bootstrapping: Each subset is created using random sampling with replacement (bootstrap=True), meaning some data points may appear multiple times in a subset while others are omitted.
- Aggregation: For classification tasks, the final prediction is made by majority voting across all individual models.
- Benefit: Primarily reduces variance and prevents overfitting.

2. Boosting (AdaBoost - Adaptive Boosting):
- Concept: Boosting trains multiple weak learners sequentially, where each new model focuses on correcting the mistakes made by the previous models.
- Weight Adjustment: Initially, all training instances are given equal weights. After each iteration, the weights of misclassified instances are increased so that the next weak learner pays more attention to harder-to-classify examples.
- Weighted Voting: Each learner is assigned a weight based on its accuracy (scaled by learning_rate), and the final prediction is determined through a weighted majority vote.
- Benefit: Primarily reduces bias and converts weak learners into a strong classifier.

Algorithm:

A. Bagging Algorithm (BaggingClassifier):
1. Load the Iris dataset and split it into training (X_train, y_train) and testing (X_test, y_test) sets.
2. Define a base estimator (DecisionTreeClassifier with max_depth=2).
3. Generate N (n_estimators=10) bootstrap subsets by randomly sampling 80% (max_samples=0.8) of the training data with replacement (bootstrap=True).
4. Train an independent Decision Tree classifier on each bootstrap subset in parallel.
5. Aggregate predictions from all 10 trees using majority voting and evaluate train and test accuracy.

B. AdaBoost Algorithm (AdaBoostClassifier):
1. Initialize equal weights w_i = 1 / m for all m training samples.
2. For each estimator t = 1 to N (n_estimators=10):
   - Train a weak Decision Tree (max_depth=2) using the current sample weights.
   - Compute the weighted error rate of the model.
   - Compute the learner's voting weight (alpha) scaled by learning_rate=0.1.
   - Increase the weights of misclassified samples and decrease the weights of correctly classified samples.
3. Combine all weak learners using weighted majority voting to make final predictions and evaluate accuracy.

Comparison Between Bagging and Boosting:
- Training Style: Bagging trains models in parallel (independent); Boosting trains models sequentially (dependent on previous model).
- Sampling Method: Bagging uses random sampling with replacement (Bootstrap); Boosting uses a weighted dataset focusing on misclassified points.
- Goal: Bagging reduces variance (prevents overfitting); Boosting reduces bias (improves weak learners).
- Final Decision: Bagging uses equal-weight majority voting; Boosting uses weighted majority voting based on model accuracy.
